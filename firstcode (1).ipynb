```python
Product_name=[]
Prices=[]
Description=[]
Reviews=[]
for i in range(2,12):
   url="https://www.flipkart.com/search?q=air+conditioner+under+50000&as=on&as-show=on&otracker=AS_Query_OnlineSuffix_3_22_na_na_na&otracker1=AS_Query_OnlineSuffix_3_22_na_na_na&as-pos=3&as-type=RECENT&suggestionId=air+conditioner+under+50000&requestId=8eb73ff8-9057-4549-8e41-4deaab8b8422&as-searchtext=airconditi+under+50000&page="+str(i)
   r=requests.get(url)
   soup=BeautifulSoup(r.text,"lxml")
   box=soup.find("div",class_="QSCKDh dLgFEE")
```

```python
   names=box.find_all("div",class_ = "RG5Slk")
   print(names)
   for i in names:
     name = i.text
     Product_name.append(name)
   print(Product_name)
```

```python
   print(len(Product_name))
```

```python
   prices = box.find_all("div" , class_="hZ3P6w DeU9vF")
   for i in prices:
     name = i.text
     Prices.append(name)
   print(Prices)
```

```python
   desc=box.find_all("ul",class_ ="HwRTzP")
   for i in desc:
     name=i.text
     Description.append(name)
   print(Description)    
```

```python
   reviews=box.find_all("div" , class_ ="MKiFS6")
   for i in reviews:
     name=i.text
     Reviews.append(name)
   print(Reviews)    
```
